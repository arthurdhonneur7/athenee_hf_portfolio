## Now doing the proper selection

```python
fund_columns = df.columns

df = df.join(ff_factors).copy()

df
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

# 1. Define your factor columns based on Screenshot 2026-05-04 at 16.42.22.png
factors = ['Mkt-RF', 'SMB', 'HML', 'RMW', 'CMA']
results_list = []

# 2. Define a helper function for Max Drawdown
def calculate_max_drawdown(returns):
    # Convert returns to a cumulative wealth index (starting at 1)
    wealth_index = (1 + returns).cumprod()
    # Calculate previous peaks
    previous_peaks = wealth_index.cummax()
    # Calculate drawdowns
    drawdowns = (wealth_index - previous_peaks) / previous_peaks
    return drawdowns.min()

# 3. Loop through the fund columns
for col in fund_columns:
    # Prepare variables
    tamp_df = df[[col] + factors].dropna()
    y = tamp_df[col]
    X = sm.add_constant(tamp_df[factors])
    
    cagr = (1 + tamp_df[col]).cumprod().iloc[-1] ** (12 / len(tamp_df)) - 1
    # Run Regression
    model = sm.OLS(y, X).fit()
    
    # Store Metrics
    results_list.append({
        'Fund': col,
        'alpha': 100 * 12 * model.params['const'],
        'beta_mkt': model.params['Mkt-RF'],
        'mean_return': 100 * tamp_df[col].mean() * 12,
        'cagr': 100 * cagr,
        'vol': 100 * np.sqrt(12)* tamp_df[col].std(),
        'sharpe_ratio': np.sqrt(12) * tamp_df[col].mean()/tamp_df[col].std(),
        't_stat_alpha': model.tvalues['const'],
        'calmar': np.abs(cagr/calculate_max_drawdown(tamp_df[col])),
        'r2': model.rsquared,
        'max_drawdown': 100*calculate_max_drawdown(tamp_df[col]),
        'alpha/vol': np.sqrt(12) * model.params['const'] / tamp_df[col].std(),
        'track_record':len(tamp_df)
    })

# 4. Create the final summary DataFrame
summary_df = pd.DataFrame(results_list).set_index('Fund')

summary_df
```

```python
t_tresh = 1.5

summary_df[summary_df['t_stat_alpha'] >= t_tresh][['alpha', 'beta_mkt', 'cagr', 'vol', 'sharpe_ratio', 't_stat_alpha']]
```

```python
list_of_candidates_str = summary_df[summary_df['t_stat_alpha'] >= t_tresh].index.astype(str).tolist()
list_of_candidates_int = summary_df[summary_df['t_stat_alpha'] >= t_tresh].index.astype(int).tolist()

df_candidates = df[list_of_candidates_str].copy()

df_candidates.head()
```
