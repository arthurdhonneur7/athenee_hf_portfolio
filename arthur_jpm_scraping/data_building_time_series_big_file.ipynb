```python
import pandas as pd
import ast

# 1. Convert the string representation of a dict into an actual Python dict
# We use ast.literal_eval because it's safer than eval() and handles single quotes
series = df['returns_series'].dropna().apply(ast.literal_eval)

# 1. Turn dict items into list of tuples: [('date1', val1), ('date2', val2)]
df_long = series.apply(lambda x: list(x.items())).to_frame('pairs')

# 2. Explode the list so each tuple gets its own row
df_long = df_long.explode('pairs')

# 3. Split the tuples into two columns
df_long[['date', 'return']] = pd.DataFrame(df_long['pairs'].tolist(), index=df_long.index)

# 4. Clean up
df_long['date'] = pd.to_datetime(df_long['date'])
df_long = df_long.drop(columns='pairs')

# 5. Join back the 'id' column from the original data
df_long = df[['id']].join(df_long).reset_index(drop=True)
```

```python
# 1. Pivot the data: Index becomes the Date, Columns become the IDs
df_final = df_long.pivot(index='date', columns='id', values='return')

# 2. (Highly Recommended) Sort the index to ensure dates are chronological
df_final = df_final.sort_index()

df_final
```

```python
df_final.columns
```

```python
import pandas as pd

# If you want to keep the index as proper Datetime objects (Recommended for plotting/analysis)
df_final.index = pd.to_datetime(df_final.index).to_period('M').to_timestamp()

# OR, if you strictly want them as strings in exactly the 'yyyy-mm-01' format
# df_final.index = pd.to_datetime(df_final.index).strftime('%Y-%m-01')

df_final
```

```python
# 1. Drop the row where the index is NaT
df_final = df_final[df_final.index.notna()]

# 2. Clean up the columns
# If your columns have a name (like 'Date' sitting awkwardly above the columns), this removes it:
df_final.columns.name = None

# If your columns are a MultiIndex (e.g., from a pivot_table) and you just want the bottom numbers:
if isinstance(df_final.columns, pd.MultiIndex):
    df_final.columns = df_final.columns.get_level_values(-1)

df_final
```

```python
(df_final/100).to_csv('./data/time_series_jpm.csv', index=True)
```
